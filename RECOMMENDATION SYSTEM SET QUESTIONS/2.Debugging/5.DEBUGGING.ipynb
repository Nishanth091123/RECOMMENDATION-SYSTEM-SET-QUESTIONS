```python
import pandas as pd

users = pd.DataFrame({
    'userId': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

ratings = pd.DataFrame({
    'user': [1, 2, 3],
    'movieId': [101, 102, 103],
    'rating': [5, 4, 3]
})

merged = pd.merge(
    users,
    ratings,
    left_on='userId',
    right_on='user'
)

print(merged)
```

```python
#8.
import pandas as pd

df = pd.DataFrame({
    'gender': ['Male', 'Female', 'Female', 'Male'],
    'education': ['Bachelors', 'Masters', 'PhD', 'Bachelors'],
    'score': [88, 92, 95, 70]
})
from sklearn.feature_selection import SelectKBest, f_classif

X = df[['gender', 'education']]
y = df['score']

selector = SelectKBest(score_func=f_classif, k=1)
X_new = selector.fit_transform(X, y)
```

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression

df = pd.DataFrame({
    'gender': ['Male', 'Female', 'Female', 'Male'],
    'education': ['Bachelors', 'Masters', 'PhD', 'Bachelors'],
    'score': [88, 92, 95, 70]
})

X = pd.get_dummies(df[['gender', 'education']], dtype=int)
y = df['score']

selector = SelectKBest(score_func=f_regression, k=1)
X_new = selector.fit_transform(X, y)

print("Selected Feature:", X.columns[selector.get_support()].tolist())
print("Shape:", X_new.shape)
```

```python
#10.
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'f1': [2.5, 0.5, 2.2, 1.9, np.nan],
    'f2': [2.4, 0.7, 2.9, 2.2, 2.5]
})
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
result = pca.fit_transform(df)
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA

df = pd.DataFrame({
    'f1': [2.5, 0.5, 2.2, 1.9, np.nan],
    'f2': [2.4, 0.7, 2.9, 2.2, 2.5]
})

# Handle missing values
df['f1'] = df['f1'].fillna(df['f1'].mean())

# Apply PCA
pca = PCA(n_components=2)
result = pca.fit_transform(df)

print(result)
print("PCA Shape:", result.shape)
```
