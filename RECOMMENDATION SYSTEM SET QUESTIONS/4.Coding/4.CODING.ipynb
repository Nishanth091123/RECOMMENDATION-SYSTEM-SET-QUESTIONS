```python

import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Create DataFrame
data = {
    'User': ['A', 'B', 'C'],
    'Book1': [5, 4, 0],
    'Book2': [3, 0, 4],
    'Book3': [0, 4, 5]
}

df = pd.DataFrame(data)

# Step 2: Create User-Book Matrix
user_book_matrix = df.set_index('User')

# Step 3: Calculate Cosine Similarity
similarity = cosine_similarity(user_book_matrix)

similarity_df = pd.DataFrame(
    similarity,
    index=user_book_matrix.index,
    columns=user_book_matrix.index
)

# Step 4: Find Similar Users for User A
similar_users = similarity_df.loc['A'].drop('A')
similar_users = similar_users.sort_values(ascending=False)

# Step 5: Find Unread Books
unread_books = user_book_matrix.columns[
    user_book_matrix.loc['A'] == 0
]

# Step 6: Recommend Books
recommended_books = set()

for user in similar_users.index:
    for book in unread_books:
        if user_book_matrix.loc[user, book] > 0:
            recommended_books.add(book)

# Step 7: Display Result
print("Recommended books for A:", recommended_books)
```

```python
#2.
movies = pd.DataFrame({
    'Title': ['A', 'B', 'C'],
    'Description': ['Hero fights villains', 'Romantic comedy film', 'Action packed thriller']
})

Question:
Recommend movies similar to Movie A based on description.

Expected Output:
Most similar movie to A: C
```

```python

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Step 1: Create Dataset
movies = pd.DataFrame({
    'Title': ['A', 'B', 'C'],
    'Description': [
        'Hero fights villains',
        'Romantic comedy film',
        'Action packed thriller'
    ]
})

# Step 2: Convert text into numerical features
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(movies['Description'])

# Step 3: Calculate cosine similarity
similarity = cosine_similarity(tfidf_matrix)

# Step 4: Find similarity for Movie A
movie_index = movies.index[movies['Title'] == 'A'][0]

similarity_scores = list(enumerate(similarity[movie_index]))

# Step 5: Exclude Movie A and sort
similarity_scores = sorted(
    [(i, score) for i, score in similarity_scores
     if i != movie_index],
    key=lambda x: x[1],
    reverse=True
)

# Step 6: Recommend most similar movie
recommended_index = similarity_scores[0][0]

print("Most similar movie to A:",
      movies.iloc[recommended_index]['Title'])
```

```python

from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Step 1: Load Iris dataset
iris = load_iris()
X = iris.data

# Step 2: Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Step 4: Print first 5 rows
print("First 5 rows after PCA transformation:")
print(X_pca[:5])
```

```python
#4.
Question:
Fit PCA on the Wine dataset and print the explained variance ratio of the first 3 components.

Expected Output:
Three values (variance ratios) adding up to less than or equal to 1.
```

```python

from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Step 1: Load Wine dataset
wine = load_wine()
X = wine.data

# Step 2: Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Apply PCA with 3 components
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

# Step 4: Print explained variance ratio
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Step 5: Print total explained variance
print("Total Explained Variance:",
      pca.explained_variance_ratio_.sum())
```

```python
#5.
Question:
Select the top 2 features from Iris dataset using Mutual Information.

Expected Output:
Top 2 features: [2 3]
```

```python

import numpy as np
from sklearn.datasets import load_iris
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# Step 1: Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Step 2: Select top 2 features using Mutual Information
selector = SelectKBest(
    score_func=lambda X, y: mutual_info_classif(
        X, y, random_state=42
    ),
    k=2
)

# Step 3: Fit the selector
selector.fit(X, y)

# Step 4: Get selected feature indices
selected_features = np.where(selector.get_support())[0]

# Step 5: Print result
print("Top 2 features:", selected_features)
```
