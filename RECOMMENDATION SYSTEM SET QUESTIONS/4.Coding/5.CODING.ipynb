```python
#1.
Question:
Use TruncatedSVD to reduce the Digits dataset to 5 components. Show shape of reduced data.

Expected Output:
(1797, 5)
```

```python

from sklearn.datasets import load_digits
from sklearn.decomposition import TruncatedSVD

# Step 1: Load Digits dataset
digits = load_digits()
X = digits.data

# Step 2: Apply TruncatedSVD
svd = TruncatedSVD(n_components=5, random_state=42)

# Step 3: Reduce dimensionality
X_reduced = svd.fit_transform(X)

# Step 4: Print shape
print("Shape of reduced data:", X_reduced.shape)
```

```python
#2.Build a Simple Content-Based Recommender (Manual Data)
Question:
Use cosine similarity to recommend books similar to Book1.

import pandas as pd

data = {
    'Book': ['Book1', 'Book2', 'Book3'],
    'Action': [1, 0, 1],
    'Romance': [0, 1, 0],
    'Fantasy': [1, 0, 0]
}
df = pd.DataFrame(data).set_index('Book')

Expected Output:
Recommended for Book1: ['Book3', 'Book2']
```

```python
#ans-

import pandas as pd

data = {
    'Book': ['Book1', 'Book2', 'Book3'],
    'Action': [1, 0, 1],
    'Romance': [0, 1, 0],
    'Fantasy': [1, 0, 0]
}
df = pd.DataFrame(data).set_index('Book')
from sklearn.metrics.pairwise import cosine_similarity

sim = cosine_similarity(df)
sim_scores = list(enumerate(sim[0]))
similar_books = sorted(sim_scores, key=lambda x: x[1], reverse=True)[1:]
recommended = [df.index[i] for i, _ in similar_books]
print("Recommended for Book1:", recommended)
```

```python
#3.
Question:
Print the explained variance for the first 10 PCA components of the Digits dataset.

Expected Output:
10 float values (explained variance)
```

```python
#
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(load_digits().data)
pca = PCA(n_components=10)
pca.fit(X)
print(pca.explained_variance_)
```

```python
#4. Use SelectFromModel with Logistic Regression (Iris Dataset)
Question:
Use SelectFromModel with Logistic Regression to select important features.

Expected Output:
Shape of selected features
(150, 2)
```

```python

from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SelectFromModel
import pandas as pd

data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = data.target

model = LogisticRegression(max_iter=200)
selector = SelectFromModel(model).fit(X, y)
X_selected = selector.transform(X)
print(X_selected.shape)
```

```python
#5. KNN Imputation and Feature Selection (Iris Dataset)
Question:
Introduce NaN in Iris data, use KNN Imputer, and apply feature selection.

Expected Output:
Shape after imputation and selection
(150, 2)
```

```python

import numpy as np
from sklearn.impute import KNNImputer
from sklearn.datasets import load_iris
from sklearn.feature_selection import SelectKBest, f_classif

data = load_iris()
X, y = data.data.copy(), data.target
X[::10] = np.nan

X_imputed = KNNImputer(n_neighbors=3).fit_transform(X)
X_selected = SelectKBest(f_classif, k=2).fit_transform(X_imputed, y)
print(X_selected.shape)
```
